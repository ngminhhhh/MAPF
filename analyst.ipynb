```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("data.csv")
```

```python
plt.figure()
plt.plot(df["rewards"])
plt.title("Reward over time")
plt.xlabel("Sample")
plt.ylabel("Reward")
plt.show()
```

```python
plt.figure()
plt.plot(df["loss"])
plt.title("Loss over time")
plt.xlabel("Sample")
plt.ylabel("Loss")
plt.show()
```
